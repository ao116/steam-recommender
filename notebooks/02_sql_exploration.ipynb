```python
from google.cloud import bigquery

PROJECT_ID = "steam-recommender-514518"
DATASET_ID = "steam_data"
FINAL = "interactions"
table_id = f"{PROJECT_ID}.{DATASET_ID}.{FINAL}"

client = bigquery.Client(project=PROJECT_ID)

query = f"""
    SELECT COUNT(*) AS total_interactions
    FROM `{table_id}`
"""

result = client.query(query).result()

for row in result:
    print(row.total_interactions)
```

```python
query = f"""
    SELECT
        COUNT(*) AS total_interactions,
        COUNT(DISTINCT user_id) AS unique_users,
        COUNT(DISTINCT game_id) AS unique_games,
        COUNTIF(positive_60m) AS positive_interactions
    FROM `{table_id}`
"""

result2 = client.query(query).result()

for row in result2:
    print(row)
```

```python
query = f"""
    SELECT
        user_id,
        game_id,
        COUNT(*) AS interaction_count
    FROM `{table_id}`

    group by 1, 2
    having interaction_count > 1
    limit 10
"""

result = client.query(query).result()

for row in result:
    print(row)
```

```python
query = f"""
    SELECT
        COUNT(*) AS duplicate_pairs,
        SUM(interaction_count - 1) AS extra_rows
    FROM (
        SELECT
            user_id,
            game_id,
            COUNT(*) AS interaction_count
        FROM `{table_id}`
        GROUP BY user_id, game_id
        HAVING interaction_count > 1
    )
"""

result = client.query(query).result()

for row in result:
    print(row)
```

```python
query = f"""
    SELECT
        user_id,
        game_id,
        MIN(play_time) AS min_playtime,
        MAX(play_time) AS max_playtime,
        COUNT(DISTINCT play_time) AS distinct_playtimes,
        COUNT(*) AS interaction_count
    FROM `{table_id}`
    GROUP BY user_id, game_id
    HAVING distinct_playtimes > 1
    LIMIT 10
"""

result = client.query(query).result()

for row in result:
    print(row)
```

```python
query = f"""
    SELECT
        user_id,
        game_id,
        MAX(play_time) AS play_time,
        MAX(play_time) >= 60 AS positive_60m
    FROM `{table_id}`
    group by 1, 2
    LIMIT 10
"""

result = client.query(query).result()

for row in result:
    print(row)
```

```python
query = f"""
    SELECT
        COUNTIF(user_id IS NULL) AS null_user_ids,
        COUNTIF(game_id IS NULL) AS null_game_ids,
        COUNTIF(play_time IS NULL) AS null_playtimes,
        COUNTIF(positive_60m IS NULL) AS null_positive_labels,

        COUNTIF(TRIM(user_id) = '') AS empty_user_ids,
        COUNTIF(TRIM(game_id) = '') AS empty_game_ids,

        COUNTIF(play_time < 0) AS negative_playtimes,
        MIN(play_time) AS min_playtime,
        MAX(play_time) AS max_playtime,

        COUNTIF(
            positive_60m != (play_time >= 60)
        ) AS inconsistent_labels

    FROM `{table_id}`
"""

result = client.query(query).result()

for row in result:
    for key, value in row.items():
        print(f"{key}: {value}")
```

```python
clean_table_id = f"{PROJECT_ID}.{DATASET_ID}.interactions_clean"

query = f"""
    CREATE OR REPLACE TABLE `{clean_table_id}` AS
    SELECT
        user_id,
        game_id,
        MAX(play_time) AS play_time,
        MAX(play_time) >= 60 AS positive_60m
    FROM `{table_id}`
    GROUP BY user_id, game_id
"""

client.query(query).result()

print("Cleaned table created successfully")
```
